### 5.Filtering & Boolean Indexing

```python

df[df['age']>28]
```

```python
# multiple conditions-combine with & (and)/|(or)
df[(df['age']>25)&(df['city']=='NY')]
```

```python
df[(df['age']>25)|(df['city']=='NY')]
```

```python
# filter rows where a column matches a set of values
df[df['city'].isin(['NY','LA'])]
```

```python
# filter using a readable string expression
df.query('age>28 and city == "NY"')
```

```python
# keep matching values,replace others with NaN
df.where(df['age']>28)
```

```python
# where not
df.mask(df['age']>28)
```

### 7.Handling Missing Data

```python
df_na=pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
df_na
```

```python
# is.na()--boolean mask of misssing values
df_na.isna()
```

```python
#.notna()--opposite of .isna()
df_na.notna()
```

```python
#count missing values per column
df_na.isnull().sum()
```

```python
#drop rows containing any NaN
df_na.dropna()
```

```python
# drop column containing any NaN
df_na.dropna(axis=1)
```

```python
# fill missing  values with a constant
df_na.fillna(0)
```

```python
# forward  fill missing values with 1 not in 1 row
df_na.ffill()
```

```python
#backward fill missing values
df_na.bfill()
```

### 8.Sorting & Ranking

```python
df=pd.DataFrame({
    'name':['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','NY']
})
df
```

```python
df.sort_values('age')
```

```python
# sort by multiple columns
df.sort_values(['city','age'],ascending=[True,False])
```

```python
# .sort_index() sort rows by the index
df.sort_index()
```

```python
# rank values within a column
df['age'].rank()
```

```python
# top n rows by a column
df.nlargest(2,'age')
```

```python
# bottom n rows by a column
df.nsmallest(2,'age')
```

### 10. GroupBy Operations

```python
df
```

```python
g=df.groupby('city')
```

```python
g
```

```python
# number of rows in each group
g.size()
```

```python
#group then aggregates a single column
g['age'].mean()
```

```python
# agg() on a groupby multiply aggregations per group
g.agg({'age':['mean','sum']})
```

```python
# on a groupby-custom function per group
g.apply(lambda x:x['age'].max()-x['age'].min())
```

```python
add=lambda a,b:a+b
```

```python
add(4,5)
```

```python
# iterating over groupd with a for loop
for name,group in g:
    print(name)
    print(group)
```

```python
#get_group() retrive a specific group
g.get_group('NY')
```
